```python
# Push from local to remote repository
from git import Repo

def add_or_update_origin(repo: Repo, url: str, remote_name: str = "origin") -> str:
    """Add remote if missing; otherwise update its URL."""
    existing = [r.name for r in repo.remotes]
    if remote_name in existing:
        repo.git.remote("set-url", remote_name, url)
        return f"Updated remote '{remote_name}' -> {url}"
    repo.git.remote("add", remote_name, url)
    return f"Added remote '{remote_name}' -> {url}"

# Usage
repo = Repo(r"C:\\Git\\Projects\\Project3")
msg = add_or_update_origin(repo, "https://github.com/shauns11/Project3.git")
print(msg)
print(repo.git.remote("-v"))

# Push main to GitHub
repo.git.push("--set-upstream", "origin", "main")
```
